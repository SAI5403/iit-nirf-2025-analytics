```python
import pandas as pd
import matplotlib.pyplot as plt

x_df = pd.read_csv("iit_x_followers_2026-07-09.csv")

x_df["followers_count"] = pd.to_numeric(x_df["followers_count"], errors="coerce")
x_df["posts_count"] = pd.to_numeric(x_df["posts_count"], errors="coerce")

x_df.head()
```

```python
x_summary = x_df.sort_values("followers_count", ascending=False)

x_summary[
    [
        "short_name",
        "x_handle",
        "followers_count",
        "posts_count",
        "collection_date"
    ]
]
```

```python
plot_df = x_df.sort_values("followers_count", ascending=True)

plt.figure(figsize=(10, 8))
plt.barh(
    plot_df["short_name"],
    plot_df["followers_count"],
    color="#1DA1F2"
)

plt.xlabel("X/Twitter followers")
plt.ylabel("IIT")
plt.title("Official X/Twitter Followers Across IITs")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()

plt.savefig("iit_x_followers_barplot.png", dpi=300)
plt.show()
```

```python
post_df = x_df.sort_values("posts_count", ascending=True)

plt.figure(figsize=(10, 8))
plt.barh(
    post_df["short_name"],
    post_df["posts_count"],
    color="#555555"
)

plt.xlabel("Number of X/Twitter posts")
plt.ylabel("IIT")
plt.title("Official X/Twitter Posting Activity Across IITs")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()

plt.savefig("iit_x_posts_barplot.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    x_df["posts_count"],
    x_df["followers_count"],
    color="#E45756",
    s=70
)

for _, row in x_df.iterrows():
    plt.text(
        row["posts_count"],
        row["followers_count"],
        row["short_name"].replace("IIT ", ""),
        fontsize=8
    )

plt.xlabel("Number of X/Twitter posts")
plt.ylabel("X/Twitter followers")
plt.title("Posting Activity vs Followers Across IITs")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("iit_x_posts_vs_followers_scatter.png", dpi=300)
plt.show()
```

```python
top10 = x_df.sort_values("followers_count", ascending=False).head(10)
top10 = top10.sort_values("followers_count", ascending=True)

plt.figure(figsize=(9, 6))
plt.barh(
    top10["short_name"],
    top10["followers_count"],
    color="#4C78A8"
)

plt.xlabel("X/Twitter followers")
plt.ylabel("IIT")
plt.title("Top 10 IITs by Official X/Twitter Followers")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()

plt.savefig("iit_x_followers_top10.png", dpi=300)
plt.show()
```
